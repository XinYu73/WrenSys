```python
import os
from pymatgen.io.vasp import Poscar
from pymatgen.core import Structure


def check_of_disorder(struct):
    return any(list(map(lambda x: len(x.elements) > 1, struct.species_and_occu)))
```

```python
for item in os.listdir("./all_2022_2024_cif"):
    try:
        struct = Structure.from_file(f"./all_2022_2024_cif/{item}")
        if not check_of_disorder(struct):
            Poscar(struct).write_file(
                f"../experimental/{item.replace('.cif', '.vasp')}"
            )
    except Exception as e:
        print(item, e)
```
